```python
# Cell 1: Compile Final Follow-up Catalog
import pandas as pd
from datetime import datetime

# Load the validated candidates
df_validated = pd.read_csv('../results/candidates/validated_candidates.csv')
anomalies = df_validated[df_validated['known_object'] == False].copy()

# Cell 2: Format Coordinates for Telescope Pointing (Sexagesimal)
from astropy.coordinates import SkyCoord
import astropy.units as u

def dec_to_sexagesimal(ra, dec):
    c = SkyCoord(ra=ra*u.degree, dec=dec*u.degree, frame='icrs')
    return c.to_string('hmsdms', sep=':', precision=2)

anomalies['radec_sexagesimal'] = anomalies.apply(lambda row: dec_to_sexagesimal(row['ra'], row['dec']), axis=1)

# Cell 3: Select and rename columns for the final report
followup_catalog = anomalies[['source_id', 'radec_sexagesimal', 'ra', 'dec', 'score_All_Fused']].copy()
followup_catalog.rename(columns={
    'score_All_Fused': 'Anomaly_Score',
    'radec_sexagesimal': 'RA_Dec_J2000'
}, inplace=True)

# Sort by highest anomaly score
followup_catalog = followup_catalog.sort_values(by='Anomaly_Score', ascending=False)

# Cell 4: Export to CSV and LaTeX for paper inclusion
date_str = datetime.today().strftime('%Y%m%d')
export_path_csv = f'../results/final_anomaly_catalog_{date_str}.csv'
export_path_tex = f'../results/final_anomaly_catalog_{date_str}.tex'

followup_catalog.to_csv(export_path_csv, index=False)
followup_catalog.head(20).to_latex(export_path_tex, index=False, caption='Top 20 Unclassified Anomalous Candidates for Follow-up', label='tab:anomalies')

print(f"Successfully generated follow-up catalog with {len(followup_catalog)} targets.")
display(followup_catalog.head())
```
