```python
# notebooks/03_data_quality.ipynb snippet
import pandas as pd

gaia_df = pd.read_csv('../data/raw/gaia_50k.csv')

# Calculate Parallax SNR
gaia_df['parallax_snr'] = gaia_df['parallax'] / gaia_df['parallax_error']

# Quality Filter: Retain objects with reliable distance measurements
clean_df = gaia_df[gaia_df['parallax_snr'] >= 5].copy()
clean_df.to_csv('../data/intermediate/gaia_clean.csv', index=False)
```
