```python
import pandas as pd
import os
from astroquery.gaia import Gaia

# Create the data directory if it doesn't exist
os.makedirs('../data/raw', exist_ok=True)

print("Launching query to Gaia Archive DR3...")

# We select 50,000 objects. 
# WHERE parallax > 1 ensures we are looking at objects within ~1000 parsecs (eliminating distant noise/galaxies).
# bp_rp IS NOT NULL ensures we have colour data for our baseline.
query = """
SELECT TOP 50000
    source_id, ra, dec, parallax, parallax_error, 
    pmra, pmdec, phot_g_mean_mag, bp_rp
FROM gaiadr3.gaia_source
WHERE parallax > 1 
AND bp_rp IS NOT NULL
"""

# Asynchronous query is standard practice for astronomical databases
job = Gaia.launch_job_async(query)
results = job.get_results()

# Convert to Pandas and save
gaia_df = results.to_pandas()
output_path = '../data/raw/gaia_50k.csv'
gaia_df.to_csv(output_path, index=False)

print(f"Successfully downloaded {len(gaia_df)} objects!")
print(f"Data saved to {output_path}")
```
